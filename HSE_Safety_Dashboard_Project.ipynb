# 1. Import Libraries

The project uses pandas for data handling, Plotly for interactive visualisations, and Streamlit for dashboard development.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import streamlit as st
from datetime import datetime
```

# 2. Load Dataset

The dashboard uses Streamlit caching because safety datasets can contain thousands of records. Caching prevents unnecessary reloads when users interact with dashboard filters.

```python
@st.cache_data
def load_data():
    df = pd.read_csv("data/OSHA_Safety_Incidents.csv")
    return df

df = load_data()

df.head()
```

# 3. Data Cleaning and Preparation

Before analysis, the data is prepared by:
- Converting incident dates into date format
- Removing duplicate records
- Creating time-based fields
- Creating severity indicators

Clean data improves accuracy of safety reporting.

```python
df['Event Date'] = pd.to_datetime(
    df['Event Date'],
    errors='coerce'
)

df['Year'] = df['Event Date'].dt.year
df['Month'] = df['Event Date'].dt.month
df['Day of Week'] = df['Event Date'].dt.day_name()

df.info()
```

# 4. Exploratory Safety Analysis

Initial analysis identifies important safety questions:

- Are incidents increasing or decreasing?
- Which incident categories occur most often?
- Which locations require additional attention?
- How many incidents represent critical safety concerns?

```python
annual_incidents = (
    df.groupby('Year')
    .size()
    .reset_index(name='Incident Count')
)

fig = px.line(
    annual_incidents,
    x='Year',
    y='Incident Count',
    title='Workplace Incident Trend'
)

fig.show()
```
